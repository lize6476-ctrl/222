# 🎯 Система рекомендации хобби для студентов
### Hobby Recommendation System

Программа задаёт вопросы о твоих интересах и рекомендует **топ-3 хобби**, подходящих именно тебе.

**Используемые темы:**
- `if / elif / else` — логические условия
- `while` — цикл
- `for`, `range`, списки, кортежи
- Методы строк и списков
- Словари (`dict`) и множества (`set`)
- Вложенные структуры, `sorted()`
- Функции (`def`)

```python
# ============================================================
#  БАЗА ХОББИ  (dict со вложенными структурами)
#  Каждое хобби имеет: описание, теги-интересов, уровень активности
# ============================================================

HOBBIES = {
    "Рисование": {
        "description": "Создавай картины, иллюстрации и скетчи",
        "tags": {"творчество", "спокойствие", "искусство", "дома"},
        "activity": "низкая",
        "cost": "низкая"
    },
    "Фотография": {
        "description": "Запечатлевай красоту мира вокруг тебя",
        "tags": {"творчество", "прогулки", "искусство", "технологии"},
        "activity": "средняя",
        "cost": "средняя"
    },
    "Программирование": {
        "description": "Создавай игры, приложения и сайты",
        "tags": {"технологии", "логика", "дома", "карьера"},
        "activity": "низкая",
        "cost": "низкая"
    },
    "Музыка": {
        "description": "Учись играть на гитаре, пианино или петь",
        "tags": {"творчество", "искусство", "дома", "общение"},
        "activity": "низкая",
        "cost": "средняя"
    },
    "Скалолазание": {
        "description": "Покоряй стены и природные скалы",
        "tags": {"спорт", "адреналин", "природа", "общение"},
        "activity": "высокая",
        "cost": "средняя"
    },
    "Велоспорт": {
        "description": "Исследуй город и природу на велосипеде",
        "tags": {"спорт", "природа", "прогулки", "здоровье"},
        "activity": "высокая",
        "cost": "средняя"
    },
    "Кулинария": {
        "description": "Готовь вкусные блюда из разных кухонь мира",
        "tags": {"творчество", "дома", "общение", "здоровье"},
        "activity": "низкая",
        "cost": "низкая"
    },
    "Чтение": {
        "description": "Погружайся в новые миры через книги",
        "tags": {"спокойствие", "логика", "дома", "карьера"},
        "activity": "низкая",
        "cost": "низкая"
    },
    "Танцы": {
        "description": "Выражай себя через движение и ритм",
        "tags": {"творчество", "спорт", "общение", "искусство"},
        "activity": "высокая",
        "cost": "низкая"
    },
    "Йога": {
        "description": "Развивай гибкость тела и спокойствие ума",
        "tags": {"спорт", "спокойствие", "дома", "здоровье"},
        "activity": "средняя",
        "cost": "низкая"
    },
    "Настольные игры": {
        "description": "Стратегии, ролевые игры и головоломки с друзьями",
        "tags": {"логика", "общение", "дома", "адреналин"},
        "activity": "низкая",
        "cost": "низкая"
    },
    "Садоводство": {
        "description": "Выращивай растения, цветы и овощи",
        "tags": {"природа", "спокойствие", "здоровье", "дома"},
        "activity": "средняя",
        "cost": "низкая"
    }
}

print("База хобби загружена:", len(HOBBIES), "вариантов")
print("Хобби:", ", ".join(HOBBIES.keys()))
```

```python
# ============================================================
#  ВСПОМОГАТЕЛЬНЫЕ ФУНКЦИИ  (def, if/elif/else, for, set)
# ============================================================

def get_valid_answer(prompt, valid_options):
    """
    Запрашивает ввод у пользователя, пока не получит допустимый ответ.
    Использует: while, if/else, методы строк (.strip, .lower)
    """
    options_str = " / ".join(valid_options)
    while True:                              # while-цикл
        answer = input(f"{prompt} [{options_str}]: ").strip().lower()
        if answer in valid_options:          # if-условие
            return answer
        else:
            print(f"  Пожалуйста, введи один из вариантов: {options_str}")


def score_hobby(hobby_tags, user_tags, user_activity, hobby_activity, user_cost, hobby_cost):
    """
    Считает совпадение хобби с предпочтениями пользователя.
    Использует: set (пересечение), if/elif/else, арифметика
    """
    # Пересечение множеств тегов (set)
    common_tags = hobby_tags & user_tags
    tag_score = len(common_tags) * 2          # каждый тег = 2 очка

    # Бонус за совпадение активности
    if user_activity == hobby_activity:
        tag_score += 3
    elif abs(["низкая", "средняя", "высокая"].index(user_activity) -
             ["низкая", "средняя", "высокая"].index(hobby_activity)) == 1:
        tag_score += 1

    # Бонус за совпадение бюджета
    if user_cost == hobby_cost:
        tag_score += 2

    return tag_score


def print_separator(char="─", length=50):
    """Печатает разделительную линию. Использует: строковые методы."""
    print(char * length)


def display_top_hobbies(ranked_list, top_n=3):
    """
    Выводит топ-N хобби с описанием.
    Использует: for, range, tuple-распаковка, f-string
    """
    medals = ("🥇", "🥈", "🥉")              # кортеж медалей
    print_separator("=")
    print("       ТОП-3 ХОББИ ДЛЯ ТЕБЯ")
    print_separator("=")

    for i in range(min(top_n, len(ranked_list))):   # for + range
        name, score = ranked_list[i]                # распаковка кортежа
        hobby_info = HOBBIES[name]
        medal = medals[i]

        print(f"\n{medal}  {name.upper()}  (очки: {score})")
        print(f"   {hobby_info['description']}")
        print(f"   Активность: {hobby_info['activity']}  |  Бюджет: {hobby_info['cost']}")
        tags_str = ", ".join(sorted(hobby_info["tags"]))  # sorted()
        print(f"   Теги: {tags_str}")

    print_separator("=")


print("Функции определены успешно!")
```

```python
# ============================================================
#  ОСНОВНАЯ ПРОГРАММА  (главная функция)
# ============================================================

def run_hobby_quiz():
    """
    Главная функция: проводит опрос и выдаёт рекомендации.
    Использует: все изученные темы
    """

    # --- Приветствие ---
    print_separator("*")
    print("*   СИСТЕМА РЕКОМЕНДАЦИИ ХОББИ ДЛЯ СТУДЕНТОВ   *")
    print_separator("*")
    print()

    name = input("Как тебя зовут? ").strip().capitalize()  # метод строки .capitalize()
    if name == "":                                          # if-условие
        name = "Студент"
    print(f"\nПривет, {name}! Отвечай на вопросы, и я подберу хобби специально для тебя.\n")

    # --- Вопросы ---
    questions = [                                            # список вопросов
        ("Любишь ли ты творческие занятия?",
         ["да", "нет"], "творчество"),
        ("Интересуют ли тебя технологии и гаджеты?",
         ["да", "нет"], "технологии"),
        ("Тебе нравятся занятия на природе?",
         ["да", "нет"], "природа"),
        ("Ты любишь общаться и заниматься с другими людьми?",
         ["да", "нет"], "общение"),
        ("Тебе важно спокойствие и тишина?",
         ["да", "нет"], "спокойствие"),
        ("Ты хочешь заниматься искусством (музыка, рисование)?",
         ["да", "нет"], "искусство"),
        ("Нравится ли тебе логика и решение задач?",
         ["да", "нет"], "логика"),
        ("Думаешь ли о карьерных преимуществах хобби?",
         ["да", "нет"], "карьера"),
    ]

    user_tags = set()                                       # пустое множество

    print("--- Блок 1: Твои интересы ---")
    for question, options, tag in questions:                # for по списку
        answer = get_valid_answer(question, options)
        if answer == "да":                                  # if/else
            user_tags.add(tag)                              # добавление в set

    print()
    print("--- Блок 2: Твои предпочтения ---")

    activity_answer = get_valid_answer(
        "Какой уровень физической активности тебе подходит?",
        ["низкая", "средняя", "высокая"]
    )

    # Предпочтение заниматься дома
    home_answer = get_valid_answer("Предпочитаешь хобби, которым можно заниматься дома?",
                                   ["да", "нет"])
    if home_answer == "да":
        user_tags.add("дома")

    # Бюджет
    cost_answer = get_valid_answer(
        "Какой бюджет на хобби тебя устраивает?",
        ["низкая", "средняя"]
    )

    # Предпочтение спортивного хобби
    sport_answer = get_valid_answer("Хочешь ли ты, чтобы хобби было связано со спортом?",
                                    ["да", "нет"])
    if sport_answer == "да":
        user_tags.add("спорт")

    health_answer = get_valid_answer("Важно ли тебе, чтобы хобби было полезно для здоровья?",
                                     ["да", "нет"])
    if health_answer == "да":
        user_tags.add("здоровье")

    adrenaline_answer = get_valid_answer("Любишь ли ты азарт и адреналин?",
                                         ["да", "нет"])
    if adrenaline_answer == "да":
        user_tags.add("адреналин")

    # --- Подсчёт очков ---
    print("\nАнализирую твои ответы...")

    scores = {}                                              # словарь для очков

    for hobby_name, hobby_data in HOBBIES.items():          # for по словарю
        s = score_hobby(
            hobby_data["tags"],
            user_tags,
            activity_answer,
            hobby_data["activity"],
            cost_answer,
            hobby_data["cost"]
        )
        scores[hobby_name] = s

    # Сортировка по убыванию очков (sorted + lambda)
    ranked = sorted(scores.items(), key=lambda x: x[1], reverse=True)

    # --- Вывод результата ---
    print(f"\nТвои теги интересов: {', '.join(sorted(user_tags))}")
    display_top_hobbies(ranked, top_n=3)

    # --- Итоговое сообщение ---
    top_hobby = ranked[0][0]                                 # название лучшего хобби
    print(f"\nМой совет, {name}: начни с '{top_hobby}'.")
    print("Это хобби лучше всего соответствует твоим интересам!")
    print()

    # Ещё раз запустить?
    again = get_valid_answer("Хочешь пройти тест ещё раз?", ["да", "нет"])
    if again == "да":
        print("\n" + "=" * 50 + "\n")
        run_hobby_quiz()                                     # рекурсивный вызов
    else:
        print("\nСпасибо за участие! Удачи в новом хобби! 🎉")


print("Программа готова. Запусти следующую ячейку чтобы начать!")
```

```python
# ============================================================
#  ЗАПУСК ПРОГРАММЫ
# ============================================================

run_hobby_quiz()
```

---
## Использованные темы Python

| Тема | Где используется |
|------|------------------|
| `if / elif / else` | `score_hobby()`, `run_hobby_quiz()`, проверка ввода |
| `while` | `get_valid_answer()` — цикл до правильного ввода |
| `for` + `range` | Обход вопросов, хобби, вывод топ-3 |
| Списки / Кортежи | `questions` — список, `medals` — кортеж |
| Методы строк | `.strip()`, `.lower()`, `.capitalize()`, `.join()` |
| Методы списков | `.append()` через `set.add()` |
| `set` (множество) | `user_tags`, пересечение `&`, `sorted()` |
| `dict` (словарь) | `HOBBIES`, `scores` — вложенные словари |
| Вложенные структуры | `HOBBIES[name]["tags"]` — словарь внутри словаря |
| `sorted()` | Сортировка хобби по очкам |
| Функции `def` | 5 функций: `get_valid_answer`, `score_hobby`, `print_separator`, `display_top_hobbies`, `run_hobby_quiz` |
