# Hobby Recommendation System
Answer a few questions and get your **top-3 hobby** suggestions!

```python
# --- Data: hobbies with tags, activity level, cost ---
HOBBIES = {
    "Drawing":        {"tags": {"creative", "calm", "indoors"},          "activity": "low",    "cost": "low"},
    "Photography":    {"tags": {"creative", "outdoors", "tech"},          "activity": "medium", "cost": "medium"},
    "Coding":         {"tags": {"tech", "logic", "indoors"},              "activity": "low",    "cost": "low"},
    "Music":          {"tags": {"creative", "social", "indoors"},         "activity": "low",    "cost": "medium"},
    "Cycling":        {"tags": {"sport", "outdoors", "health"},           "activity": "high",   "cost": "medium"},
    "Cooking":        {"tags": {"creative", "indoors", "health"},         "activity": "low",    "cost": "low"},
    "Reading":        {"tags": {"calm", "logic", "indoors"},              "activity": "low",    "cost": "low"},
    "Dancing":        {"tags": {"creative", "sport", "social"},           "activity": "high",   "cost": "low"},
    "Yoga":           {"tags": {"sport", "calm", "health", "indoors"},    "activity": "medium", "cost": "low"},
    "Board Games":    {"tags": {"logic", "social", "indoors"},            "activity": "low",    "cost": "low"},
}

# --- Questions: (text, tag awarded on 'yes') ---
QUESTIONS = [
    ("Do you enjoy creative activities?",        "creative"),
    ("Are you interested in technology?",         "tech"),
    ("Do you prefer outdoor activities?",         "outdoors"),
    ("Do you like socializing with people?",      "social"),
    ("Do you value peace and quiet?",             "calm"),
    ("Do you want a sport-related hobby?",        "sport"),
    ("Is health improvement important to you?",   "health"),
    ("Do you enjoy puzzles and logic?",           "logic"),
    ("Do you prefer staying indoors?",            "indoors"),
]


def ask(prompt, options):
    """Ask until a valid option is entered (while loop)."""
    while True:
        ans = input(f"{prompt} ({'/'.join(options)}): ").strip().lower()
        if ans in options:
            return ans
        print("  Please choose:", "/".join(options))


def score(hobby, user_tags, user_activity, user_cost):
    """Score a hobby: tag matches + activity + cost bonuses."""
    levels = ["low", "medium", "high"]
    points = len(hobby["tags"] & user_tags) * 2          # set intersection
    points += 3 if hobby["activity"] == user_activity else 0
    points += 2 if hobby["cost"] == user_cost else 0
    return points


def run():
    print("=" * 45)
    print("    HOBBY RECOMMENDATION SYSTEM")
    print("=" * 45)

    name = input("Your name: ").strip().capitalize() or "Student"
    print(f"\nHi, {name}! Answer yes/no questions.\n")

    # Collect user tags (set)
    user_tags = set()
    for question, tag in QUESTIONS:                       # for loop over list of tuples
        if ask(question, ["yes", "no"]) == "yes":         # if/else
            user_tags.add(tag)

    user_activity = ask("\nPreferred activity level?", ["low", "medium", "high"])
    user_cost     = ask("Preferred budget?",           ["low", "medium"])

    # Score and rank all hobbies (dict, sorted)
    scores = {name: score(data, user_tags, user_activity, user_cost)
              for name, data in HOBBIES.items()}
    ranked = sorted(scores.items(), key=lambda x: x[1], reverse=True)

    # Display top-3 (tuple of medals, for + range)
    medals = ("1st", "2nd", "3rd")
    print("\n" + "=" * 45)
    print(f"  YOUR TOP-3 HOBBIES, {name.upper()}:")
    print("=" * 45)
    for i in range(3):                                    # range
        hobby_name, pts = ranked[i]                       # tuple unpacking
        tags_str = ", ".join(sorted(HOBBIES[hobby_name]["tags"]))  # sorted + join
        print(f"  {medals[i]}: {hobby_name} (score: {pts})")
        print(f"      Tags: {tags_str}")
    print("=" * 45)
    print(f"\nSuggestion: start with '{ranked[0][0]}'!")


run()
```
